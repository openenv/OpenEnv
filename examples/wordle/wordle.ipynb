## 1. Play the environment

The environment runs as a Space. Each guess returns the feedback for every letter: `G` (right letter, right spot), `Y` (in the word, wrong spot) and `X` (not in the word).

```python
from textarena_env import TextArenaAction, TextArenaEnv

with TextArenaEnv(base_url="https://openenv-wordle.hf.space").sync() as env:
    result = env.reset()
    print(result.observation.messages[0].content)
    result = env.step(TextArenaAction(message="[crane]"))
    print(result.observation.messages[0].content[-200:])
```

## 2. Generate the SFT data

A solver plays 1,000 games against the same environment class the Space runs (in-process, so it takes seconds) and writes each one as a tool-calling conversation, with a short reasoning before every guess. Enter your Hugging Face username or organization below. Set `env_url` to the URL of your duplicated Wordle Space before starting GRPO.

```python
import subprocess
import sys

hub_namespace = input("Hugging Face username or organization: ").strip()
if not hub_namespace or any(c not in "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789-_" for c in hub_namespace):
    raise ValueError("Enter a valid Hub username or organization")
env_url = input("URL of your duplicated Wordle Space: ").strip()
if not env_url.startswith("https://") or any(c.isspace() for c in env_url) or any(c in env_url for c in "<>"):
    raise ValueError("Enter your Space's HTTPS URL")
dataset_id = f"{hub_namespace}/wordle-tool-calls-sft"
sft_model_id = f"{hub_namespace}/Qwen3-1.7B-Wordle-SFT"
grpo_model_id = f"{hub_namespace}/Qwen3-1.7B-Wordle-SFT-GRPO"

!wget -q https://raw.githubusercontent.com/openenv/OpenEnv/main/examples/wordle/generate_sft_data.py https://raw.githubusercontent.com/openenv/OpenEnv/main/examples/wordle/sft.py https://raw.githubusercontent.com/openenv/OpenEnv/main/examples/wordle/grpo.py https://raw.githubusercontent.com/openenv/OpenEnv/main/examples/wordle/evaluate.py https://raw.githubusercontent.com/openenv/OpenEnv/main/examples/wordle/make_gif.py https://raw.githubusercontent.com/openenv/OpenEnv/main/examples/wordle/secrets30.txt
subprocess.run([sys.executable, "generate_sft_data.py", "--games", "1000", "--push-to-hub", dataset_id], check=True)
```

## 3. SFT and GRPO on Hugging Face Jobs

Both steps run on one A100 80GB (`a100-large`, $2.50/h). The SFT warmup takes 7 min and GRPO 2 h 15 min, about $6 in total.

GRPO plays 64 games at once, one environment session each. [Duplicate the `openenv/wordle` Space](https://huggingface.co/spaces/openenv/wordle?duplicate=true), set the `MAX_CONCURRENT_ENVS=64` variable on your copy and pass its URL as `--env-url`.

```python
subprocess.run([
    "hf", "jobs", "uv", "run", "--flavor", "a100-large", "--timeout", "1h",
    "--secrets", "HF_TOKEN", "sft.py",
    "--dataset", dataset_id, "--hub-model-id", sft_model_id,
], check=True)
```

Once the SFT job finishes, start GRPO from the SFT model. Follow the run with `hf jobs logs <job-id>` and on Trackio.

```python
subprocess.run([
    "hf", "jobs", "uv", "run", "--flavor", "a100-large", "--timeout", "4h",
    "--secrets", "HF_TOKEN", "grpo.py",
    "--model", sft_model_id, "--hub-model-id", grpo_model_id,
    "--env-url", env_url,
], check=True)
```

## 4. Compare the models

`evaluate.py` plays a model greedily and saves every game. `--secrets` runs the environment in-process with fixed secret words, so every model plays the same ones. This uses the published models, swap in yours once they're trained. The cells below play the first 10 of the 30 words. On a laptop with Apple silicon, all 30 take 3 to 10 minutes per model. The evaluator allows six legal guesses (up to 12 submissions including invalid warnings) and stops when the environment ends the game. The tutorial’s published scores used the earlier six-submission cap and have not been recomputed with this budget.

```python
!python evaluate.py --model Qwen/Qwen3-1.7B --max-new-tokens 256 --secrets "$(head -c 59 secrets30.txt)" --out base.jsonl
!python evaluate.py --model openenv/Qwen3-1.7B-Wordle-SFT --thinking --max-new-tokens 512 --secrets "$(head -c 59 secrets30.txt)" --out sft.jsonl
!python evaluate.py --model openenv/Qwen3-1.7B-Wordle-SFT-GRPO --thinking --max-new-tokens 512 --secrets "$(head -c 59 secrets30.txt)" --out grpo.jsonl
```
